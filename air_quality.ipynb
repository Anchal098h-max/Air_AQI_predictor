```python
df["No-grp"]  = df["NH3"] *df["NOx"]
df["Particales"]  = df["PM2.5"]* (df["PM10"])
df["NO2_SO2"] = (df["NO2"])+ df["SO2"]
```

```python
df["VOC"] = df["Benzene"]+ df["Toluene"]+ df["Xylene"]
df["Pollution_Load"] = ( df["PM2.5"] + df["PM10"] + df["NOx"])
df["Gas_Total"] = (df["NO"] +df["NO2"] +df["NOx"] +df["SO2"] + df["CO"])
```

```python
from sklearn.model_selection import train_test_split
x = df.drop( 'AQI' , axis = 1)
y = df["AQI"]
```

```python
x_train , x_test , y_train ,  y_test = train_test_split( x , y,  test_size= 0.2 , random_state= 42)
```

```python
df.head()
```

```python
from  sklearn.pipeline import Pipeline 
from sklearn.compose import  ColumnTransformer 
from sklearn.impute import SimpleImputer
from sklearn.preprocessing  import OneHotEncoder  , StandardScaler   
```

```python
df.info()
```

```python
num_col  =  x_train.select_dtypes(include= ["number"]).columns
cat_cols = x_train.select_dtypes(include= ["object"]).columns
```

```python
num_pipeline = Pipeline( [
     ("imputer" , SimpleImputer( strategy= "median")) , 
     #  ( "Robust" , RobustScaler()), 
     ( "scaler" , StandardScaler())
])
cat_pipeline = Pipeline(
     [
         ( " cat_imputer" ,  SimpleImputer( strategy= "most_frequent")) , 
         ( "encoder" , OneHotEncoder( handle_unknown= "ignore"))
     ]
)
preprocessor = ColumnTransformer (
     [
          ("num" , num_pipeline , num_col) , 
          ("cat" , cat_pipeline , cat_cols)
     ]
)
x_train  = preprocessor.fit_transform(x_train)
x_test = preprocessor.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit( x_train , y_train)
y_pred = lr.predict(x_test)
```

```python
plt.figure( figsize=( 8 , 6))
sns.scatterplot(
    x = y_test, 
    y = y_pred, 
    alpha = 0.6
)
plt.plot(
    [y_test.min() , y_test.max()] , 
    [ y_test.min() , y_test.max()] , 
    color = "red" , 
    linewidth = 2 
)
plt.show()
```

```python
from sklearn.metrics import r2_score , mean_absolute_error , root_mean_squared_error  , mean_absolute_percentage_error
```

```python
print( r2_score ( y_test , y_pred))
print( mean_absolute_error ( y_test , y_pred))
print( root_mean_squared_error ( y_test , y_pred))
```

```python
residual  = y_test - y_pred
plt.figure( figsize=( 8 , 8 ))
plt.scatter( y_pred , residual , alpha= 0.6)
plt.axhline( 0 , color = "red") 
```
